### EDA - Exploratory Data Analysis
- EDA is the first and most critical step in Data Analytics
- It helps you understand what your data is saying before making decisions or building models

```python
### import Basic libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel('flight_price.xlsx')
```

```python
df.head()
```

```python
### Get Basic Information about data
df.info()
```

```python
df.shape
```

```python
df.isnull()
```

```python
### Feature Engineering
df['Date']=df['Date_of_Journey'].str.split('/').str[0]
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head(2)
```

```python
df.info()
```

```python
## converting the data type into integer
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
### Drop the Date of Journey Column
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.head()
```

```python
## this is used to remove the extra data from the arrival time
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])

df['Arrival_Hour']=df['Arrival_Time'].str.split(':').str[0]
df['Arrival_Min']=df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df['Arrival_Hour']=df['Arrival_Hour'].astype(int)
df['Arrival_Min']=df['Arrival_Min'].astype(int)
```

```python
df.info()
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python

df['Dep_Hour']=df['Dep_Time'].str.split(':').str[0]
df['Dep_Min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df['Dep_Hour']=df['Dep_Hour'].astype(int)
df['Dep_Min']=df['Dep_Min'].astype(int)
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df.info()
```

```python
df.head(2)
df['Total_Stops']
```

```python
df['Total_Stops']
```

```python
df['Total_Stops'].mode()[0]
```

```python
## We are making the values in the total stops
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stop':2,'3 stop':3,'4 stop':4,np.nan:1})
```

```python
df.head()
```

```python
df['Duration_Hour']=df['Duration'].str.split(' ').str[0].str.split('h').str[0]
df['Duration_Min']=df['Duration'].str.split(' ').str[1].str.split('m').str[0]
```
