#### 5 Number Summary - To Identify Outliers

```python
### MIN,Q1,Medain,Q3,MAX
```

```python
import numpy as np
```

```python
list_marks=[20,34,11,4,55,29,43,55,88,99,10,77,22,11,4,57,91,29,19,30]
```

```python
minimum,Q1,median,Q3,maximum=np.quantile(list_marks,[0,0.25,0.50,0.75,1.0])
```

```python
print(minimum)
print(Q1)
print(median)
print(Q3)
print(maximum)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*IQR
higher_fence=Q3+1.5*IQR
```

```python
print(lower_fence)
print(higher_fence)
```
